```python
import pandas as pd
import numpy as np
import difflib
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
movies_data=pd.read_csv('/content/movies.csv')
```

```python
movies_data.shape
```

```python
movies_data.head()
```

```python
selected_features=['genres','keywords','tagline','title','director']
```

```python
print(selected_features)
```

```python
for features in selected_features:
  movies_data[features]=movies_data[features].fillna('')
```

```python
#combining all 5 datas
combined_feature=movies_data['genres']+' '+movies_data['keywords']+' '+movies_data['tagline']+' '+movies_data['title']+' '+movies_data['director']
```

```python
print(combined_feature)
```

```python
#converting the text data to featur variable
vectorizer=TfidfVectorizer()
feature_vector=vectorizer.fit_transform(combined_feature)
```

```python
print(feature_vector)
```

```python
#getting similarity score using cosine similarity
similarity=cosine_similarity(feature_vector)
```

```python
print(similarity)
```

```python
similarity.shape
```

```python
#getting the movie name from the user
movie_name=input('enter the name of the movie: ')
```

```python
#Taking all the movies in a list
list_of_movies=movies_data['title'].tolist()
```

```python
print(list_of_movies)
```

```python
#finding the closest match usinf difflib
find_close_match=difflib.get_close_matches(movie_name,list_of_movies)
```

```python
print(find_close_match)
```

```python
close_match=find_close_match[0]
print(close_match)
```

```python
#finding the index of the movie with the help of title
index_of_the_movie=movies_data[movies_data.title==close_match]['index'].values[0]
print(index_of_the_movie)
```

```python
#getting a list of similar movies
similarity_score=list(enumerate(similarity[index_of_the_movie]))
print(similarity_score)
```

```python
 #sorting movies based on similarity score
sorted_similar_movies=sorted(similarity_score,key=lambda x:x[1], reverse=True)
print(sorted_similar_movies)
```

```python
#print the name of similar movies based on index
i=1
for movies in sorted_similar_movies:
  index=movies[0]
  title_from_index=movies_data[movies_data.index==index]['title'].values[0]
  if(i<30):
    print(i,'.',title_from_index)
    i+=1
```

```python
movie_name=input('enter the name of the movie: ')
list_of_movies=movies_data['title'].tolist()
find_close_match=difflib.get_close_matches(movie_name,list_of_movies)
close_match=find_close_match[0]
index_of_the_movie=movies_data[movies_data.title==close_match]['index'].values[0]
similarity_score=list(enumerate(similarity[index_of_the_movie]))
sorted_similar_movies=sorted(similarity_score,key=lambda x:x[1], reverse=True)

i=1
for movies in sorted_similar_movies:
  index=movies[0]
  title_from_index=movies_data[movies_data.index==index]['title'].values[0]
  if(i<30):
    print(i,'.',title_from_index)
    i+=1
```
